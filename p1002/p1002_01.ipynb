```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
plt.plot(x,y)
```

```python
# 축 설명글, 위치, 색상

x = [1,2,3]
y = [-15,-5,30]
plt.title('꺾은선 graph')
plt.xlabel('매출액',color='red',loc='right')
plt.ylabel('월별',color='#00aa00',loc='top') # loc(위치)
plt.plot(x,y)
plt.show()

# 범례 추가
plt.plot(x,y,label='범례(매출액)')
plt.title('그래프')
plt.legend

# 그래프 간격

plt.xticks([1,1,2,1,4,1,6,2,3])
plt.yticks([-15,10,20,40,50])

plt.plot(x,y)
plt.show()
```

```python
# 축설명글 색상, 위치
x = ['1번','2번','3번']
y = [80,60,90]
plt.title('꺾은선 graph') # 상단타이틀

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축 설명 색상, 위치
plt.ylabel('월별',color='#00AA00',loc='top') # Y출 설명글 색상, 위치

#  그래프 간격조정 ( 입격값 형태로 구성)
plt.xticks([1,1.2,1.4,1.6,2,3])
plt.yticks([-15,10,20,30])

plt.plot(x,y) # 선 그래프그리기
plt.show()
```

```python
# 축설명글 색상, 위치
x = [1,2,3]
y = [-15,10,30]

# 범례추가, 범례위치
plt.plot(x,y,label='국어점수')       # 그래프 그리기 / 설정부분보다 위쪽에 배치
plt.legend(loc='upper right') # 위쪽 오른쪽에 배치, loc표시하지 않으면 선에 걸리지 않게 자동배치
# upper, center, lower / left, right, ??

plt.title('꺾은선 graph')  # 상단타이틀

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치


# 그래프 간격조정 ( 입력값 형태로 구성 )
plt.xticks([1,1.2,1.4,1.6,2,3])
plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()
```

```python
# 축설명글 색상, 위치
x = [1,2,3]
y = [-15,10,30]

# 범례추가, 범례위치
plt.plot(x,y,label='국어점수')       # 그래프 그리기 / 설정부분보다 위쪽에 배치
# upper,center,lower / left,center,right
# plt.legend(loc='upper right')   # 위쪽 오른쪽편 위치이동 선에 걸리지 않게 자동배치
plt.legend(loc=(0.5,0.5))   # 위치를 지정

plt.title('꺾은선 graph')  # 상단타이틀

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치


# 그래프 간격조정 ( 입력값 형태로 구성 )
plt.xticks([1,1.2,1.4,1.6,2,3])
plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()
```

```python
# 선 스타일
x = [1,2,3]
y = [10,5,30]

# 범례추가, 범례위치
plt.plot(x,y,linewidth=2, marker='o',ms=10,mfc='yellow', color = 'blue')    # linewidth : 선두께, marker는 자료 표식, o,x,v, markersize(마커 크기설정, ms로 축약가능), linestyles(None, :, --,-.)
# markerfacecolor(마커 색상, mfc), markeredgecolor(mec), alpha(투명도)
# color : 선색상
# plt.plot(x,y,linewidth=2, marker='v', linestyle="None")    
# linewidth : 선두께, linestyle(선형태)="None" 선이 사라짐
# plt.scatter(x,y)
# upper,center,lower / left,center,right
# plt.legend(loc='upper right')   
plt.legend(loc=(0.5,0.5))   
```
