```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['font.size'] = 15

import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/score.xlsx',index_col="지원번호")
```

```python
# 선 스타일
x = [1,2,3]
# x = ['홍길동','유관순','강감찬']
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]
colors = ['red','green','blue']
plt.title('학생성적그래프')
# color:그래프 색상변경  :색상('red'),여러개 - ['red','green','blue']
# width:막대그래프 넓이조절, alpha:투명도
bar = plt.bar(x,y,color=colors,width=0.8,alpha=0.5)
bar[0].set_hatch('/')
bar[1].set_hatch('x')
bar[2].set_hatch('..')

# grid - 격자선, 안내선
plt.grid(axis='y',linewidth=2,color='purple',linestyle='-.',alpha=0.5)
plt.grid(axis='x')

# 가로막대그래프 생성
# plt.barh(y,x)
# rotation :각도조절
# plt.xticks(x,rotation=90)
for i,v in enumerate(y):
    plt.text(x[i],y[i]+2,v,ha='center')
plt.show()
```

```python
# 선 스타일
# x = df.index
# index = [1,2,3,4,5,6,7,8]
index = np.arange(df.shape[0])
x = df['이름']
y = df['국어']
y2 = df['영어']
y3 = df['수학']
plt.figure(figsize=(10,5))
colors = ['red','green','blue']
plt.title('학생성적그래프')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)

# 바 성적을 출력하시오.
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+1,v,ha='center')
    plt.text(i,v2+1,v2,ha='center')
    plt.text(i+0.2,v3+1,v3,ha='center')


plt.legend(ncol=3)
plt.xticks(index,df['이름']) # index의 위치에 df['이름']으로 대체
plt.yticks([0,50,60,70,80,90,10,120])
plt.show()
```
