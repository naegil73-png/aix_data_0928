```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['font.size'] = 15
```

```python
# 선스타일
x = [1,2,3]
y = [10,5,30]

plt.title('꺾은선 graph') # 상단타이틀

# figure : 그래프 사이즈변경, figsize: 크기, dpi: 해상도
# facecolor : 배경색 변경

# X선 설명부분
plt.xlabel('매출액',color='red',loc='right') # X축설명글 색상, 위치
plt.ylabel('월별',color='#00aa00',loc='top') #Y축설명글 색상, 위치

# 그래프 간격조정 ( 입력값 형태로 구성 )
plt.xticks([1,1.2,1.4,1.6,2,3])
plt.yticks([-15,10,20,30]) #간격부분 조정을 하면 꺾이는 상태를 조정
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()


plt.figure(figsize=(20,5))
plt.figure(figsize=(10,5),facecolor='yellow')
plt.plot(x,y,linewidth=2,marker='o',ms=10,mfc='black',linestyle='-.',color='blue')

#--------------------------알아둘 것
# 그래프 안에 글자를 입력 : plt.text
plt.text(x[0],y[0]+0.3,2,ha='center',color='red')
for idx, value in enumerate(y):
    plt.text(x[idx],y[idx]+0.3, value, ha ='center',color = 'red') # +0.3은 숫자와 그래프의 간격을 +(x도 가능)

plt.savefig('file/aaa.png',dpi=72)
plt.savefig('file/aaa2.png',dpi=72)


plt.show()
```

```python
# 선 스타일
x = [1,2,3]
# x = ['홍길동','유관순','강감찬']
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]
colors = ['red','green','blue']
plt.title('학생성적그래프')
# color:그래프 색상변경  :색상('red'),여러개 - ['red','green','blue']
# width:막대그래프 넓이조절, alpha:투명도
bar = plt.bar(x,y,color=colors,width=0.8,alpha=0.5)
bar[0].set_hatch('/')
bar[1].set_hatch('x')
bar[2].set_hatch('..')

# grid - 격자선, 안내선
plt.grid(axis='y',linewidth=2,color='purple',linestyle='-.',alpha=0.5)
plt.grid(axis='x')

# 가로막대그래프 생성
# plt.barh(y,x)
# rotation :각도조절
# plt.xticks(x,rotation=90)
for i,v in enumerate(y):
    plt.text(x[i],y[i]+2,v,ha='center')
plt.show()
```
