```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['font.size'] = 15

import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/score.xlsx',index_col="지원번호")
```

```python
# 선스타일
x = df.index # df['이름']으로 하면, 이름으로 전환
y = df['국어']
y2 = df['영어']
y3 = df['수학']


plt.figure(figsize=(10,5))
colors = ["blue","red","green"]
plt.title('학생성적그래프')
plt.bar(x,y,label='국어',width=0.5)
plt.bar(x,y2,bottom=y, label='영어',width=0.5) # y2값에 y+y2 합한 값
plt.bar(x,y3,bottom=y+y2,label='수학',width=0.5) # y3값에 y+y2+y3 합한 값

# ncol = 3 ; 범례가 1줄로 나타남

plt.legend(ncol=3)
plt.grid(axis='y',linestyle=':',alpha=0.5)
# 국어점수
for i, (v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i,v/2,v,ha='center',va="center")
    plt.text(i,v+v2/2,v,ha='center',va="center")
    plt.text(i,v+v2+v3/2,v,ha='center',va="center")


plt.show()
```
