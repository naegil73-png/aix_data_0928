```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('file/students.xlsx',index_col='번호')
df
```

```python
# 학교별 국어,영어,수학평균을 막대그래프로 출력하시오.
school = df.groupby('학교')[['국어', '영어', '수학']].mean()
school
```

```python
x = school.index
index = np.arange(school.shape[0])
y = school['국어']
y2 = school['영어']
y3 = school['수학']
plt.title('학교별 성적그래프')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)
plt.xticks(index,school.index)
plt.yticks([0,50,80,100])
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+2,f'{v:.1f}',ha='center')
    plt.text(i,v2+2,f'{v2:.1f}',ha='center')
    plt.text(i+0.2,v3+2,f'{v3:.1f}',ha='center')
```

```python
# 학교별, 학년별 국어,영어,수학 평균을 출력하시오.
result = df.groupby(['학교', '학년'])[['국어', '영어', '수학']].mean()
```

```python
# 합계,평균,등수 컬럼을 추가하고 값을 입력해서 출력하시오.
# 합계,평균,등수
df['합계'] = df['국어']+df['영어']+df['수학']
df['평균'] = df['합계']/3
df['등수'] = df['합계'].rank(ascending=False).astype(int)
```
