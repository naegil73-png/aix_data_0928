```python
# 학생들의 국어,영어,수학 그래프, 제목은 학생성적 그래프, 범례넣고, 그래프에 숫자 표시
```

```python
import matplotlib.pyplot as plt
import matplotlib
import pandas as pd
import numpy as np

df = pd.read_excel('file/score.xlsx',index_col='지원번호')
df
```

```python
x = df['이름']
y = df['국어']
y2 = df['영어']
y3 = df['수학']

plt.rcParams['font.family'] = "Malgun Gothic"
plt.rcParams['axes.unicode_minus'] = False

index = np.arange(df.shape[0])
plt.figure(figsize=(10,5))
plt.title('학생성적그래프')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)

for index,(value,value2,value3) in enumerate(zip(y,y2,y3)):
    plt.text(index-0.2,value+1,value,ha='center')
    plt.text(index,value2+1,value2,ha='center')
    plt.text(index+0.2,value3+1,value3,ha='center')

plt.xticks(index,df['이름']) # index를 이름으로 바꾸겠다.
plt.legend()
plt.show()
```
