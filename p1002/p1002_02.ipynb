```python
import pandas as pd
import numpy as np
df = pd.read_excel('file/score.xlsx',index_col="지원번호")
```

```python
df.index
df.columns
# 키
df['키']
```

```python
# x축: index, y : 국어점수
# 그래프 범례 : 국어점수
```

```python
x = df.index
y = df['국어']
y2 = df['영어']
y3 = df['수학']
y4 = df['과학']
y5 = df['사회']
plt.plot(x,y,y2)
plt.title('학생점수')
plt.plot(x,y,label='범례(국어점수)')
plt.plot(x,y2,label='범례(영어점수)')
plt.plot(x,y3,label='범례(수학점수)')
plt.plot(x,y4,label='범례(과학점수)')
plt.plot(x,y5,label='범례(사회점수)')
plt.legend
plt.show()

# 그래프를 2종류 이상으로 설정할 수도 있음
```
