```python
import pandas as pd
import numpy as np
df = pd.read_excel("file/score.xlsx",index_col="지원번호")
```

```python
# NaN처리(결측치 처리)
# fillna : nan를 채워줌

df.fillna(' ',inplace=True) # 결측치는 공백(" ") 처리
df
```

```python
# 학교 컬럼을 모두 nan처리 :
# np.nan : numpy명령어, nan데이터를 모두 입력
# numpy는 숫자만 가능, 문자는 안됨
df['학교'] = np.nan
df
```

```python
df.fillna("없음", inplace=True)
```

```python
#  Nan 데이터 삭제 : dropna -> nan이 포함된 모든 행을 지움. axis는 방향, how는 범위
df['학교'] = df['학교'].astype('object')
df.dropna(how='any',axis="index") # how: any - 1개만 있어도 삭제 -> 5, 6번 삭제
df.dropna(how='all',axis="index") # how: all - 모두 있어야 삭제 -> 삭제 안됨
df.dropna(how="any", axis='columns') # axis는 index, columns가 있음. 전자는 인덱스 방향, rows가로 삭제, columns는 사로삭제
df
```
