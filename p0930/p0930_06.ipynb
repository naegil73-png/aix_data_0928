```python
df = pd.read_excel("file/students.xlsx",index_col="번호")
```

```python
# s가 있는 이름을 모두 출력하시오. 대문자, 소문자 모두 확인
# 명령문에 1개이면 str없이 사용, 여러개일 때 str 입력 후 사용
df[df['이름'].str.lower().str.contains('s')] 
df
```

```python
df['총 인구수'] = df['총 인구수'].str.replace(",","").astype(int) # 숫자로 변환하려면, 1개만 변환은 안됨. 컬럼 전체를 컬럼으로 전환해야 함
```
