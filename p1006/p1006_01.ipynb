```python
df = pd.read_csv('file/drinks.csv')
df
```

```python
df['total_serving'] = df['beer_servings']+df['spirit_servings']+df['wine_servings']
df
```

```python
total_s = df.groupby(['country'])['total_serving'].sum().sort_values(ascending=False)
total_s
```

```python
df = pd.DataFrame(total_s)
df
```

```python
x = df.index
y = df['total_serving']
plt.title('총주문량')
for i,v in enumerate(y):
    plt.text(i,v+0.5,v,ha='center')
plt.bar(x,y,label='총주문량')

plt.Figure(figsize=(10,5))
plt.legend()
plt.show()
```
