```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False     # - 부호처리
matplotlib.rcParams['font.family'] = 'Malgun Gothic'  # 한글처리 : 윈도우 맑은고딕 저장
matplotlib.rcParams['font.size'] = 15                 # 전체 글자폰트크기 조정
# matplotlib.rcParams['font.family'] = 'AppleGothic'    # 맥컴 애플고딕 저장
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('file/chipotle.tsv',sep='\t')
df
```

```python
# item_name으로 그룹핑해서 quantity 합계를 막대그래프로 출력
```

```python
df = df.groupby(['item_name'])['quantity'].sum().sort_values(ascending=False)
df
```

```python
df = pd.DataFrame(df)
df
```

```python
x = df.index
y = df['quantity']
plt.figure(figsize=(15,10))
plt.bar(x,y, label='주문량')
plt.xticks(x,rotation=90) # x축의 인덱스를 90도 회전
plt.title('품목별 주문량')
plt.legend()
plt.show()
```

```python
# 주문갯수가 50개 이상, 주문금액이 50달러 이상
```
