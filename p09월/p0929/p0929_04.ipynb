```python
# 파일 저장, 읽기

import pandas as pd
data = {
    '이름':['강나래','강태원','강호림','김수찬','김재욱','박동현','박혜정','승근열'],
    '학교':['신림고','신림고','신림고','신림고','신림고','디지털고','디지털고','디지털고'] ,
    '키':[197,184,168,187,188,202,188,190],
    '국어' : [90, 40, 80, 40, 15, 80, 55, 100],
    '영어' : [85, 35, 75, 60, 20, 100, 65, 85],
    '수학' : [100, 50, 70, 70, 10, 95, 45, 90],
    '과학' : [95, 55, 80, 75, 35, 85, 40, 95],
    '사회' : [85, 25, 75, 80, 10, 80, 35, 95],
    'SW특기' : ['Python', 'Java', 'Javascript', '', '', 'C', 'PYTHON', 'C#']}

df =pd.DataFrame(data, index=['1번','2번','3번','4번','5번','6번','7번','8번'])
df.index.name = '지원번호'
```

```python
# 파일 저장 : csv, txt, xlsx 형태로 저장 가능(database프로그램과 호환가능)
df.to_csv('file\\score.csv',encoding='utf-8-sig') # df자료를 csv 파일로 저장하겠다. 파일이름은 score, 엑셀에서도 읽을 수 있으려면 sig까지 입력. 안 읽으려면, utf-8까지만 입력하면 됨. python에서는 with open~형태
```

```python
# txt파일로 저장
df.to_csv("file/score.txt") # to_score까지는 같고, encoding 필요없음, 자료를 쉼표기준으로 분리
df.to_csv("file/score1.txt",sep="\t") # 이 파일은 자료를 탭 기준으로 분리, 임의의 문자 기준으로도 분리 가능
```

```python
# xlsx 파일로 저장
df.to_excel('file/score.xlsx') # 
```
