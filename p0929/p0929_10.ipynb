```python
import pandas as pd
df = pd.read_excel("file/score.xlsx",index_col="지원번호")
df
```

```python
# 인덱싱하는 방법 반드시 알아둘 것
# loc -이름 검색, iloc-index주소로 검색
# df.loc['1번'] # 이름으로 검색
# df.iloc[0] # 주소값으로 검색
# df.index # 인덱스값 출력
# df.columns # 모든 columns 명을 검색
```

```python
# 컬럼 슬라이싱은 columns을 활용
# df['학교':'영어']  # col이름으로 슬라이싱 안됨.
# df[df.columns[1:5]] # columns 슬라이싱가능

# loc 형태 - row출력
df.loc['3번']         # 1개출력
df.loc[['1번','2번']] # 2개출력
df.loc['3번','국어']  # 1개row에서 국어컬럼만 출력
df.loc['1번':'4번']   # 1번~4번까지 출력
df.loc['1번':'4번','학교':'영어'] # index슬라이싱, column슬라이싱
df.loc['1번':'4번',['학교','영어']] # index슬라이싱, column2개는 리스트
df.loc[['1번','4번'],['학교','영어']] # index2개는리스트, column슬라이싱
```

```python
# iloc 형태 - row출력
df.loc['1번']
df.loc['1번':'4번']
df.iloc[0]
df.iloc[0:4] # 슬라이싱은 리스트 1개 가능
df.iloc[[0,2,4]] # 2개는 리스트
df.iloc[0:3,1:5] # index슬라이싱, columns슬라이싱
df.iloc[[0,2],1:5] # 2개는 리스트
# df.iloc[[0,2],[1,3,5]] # index 2개는 리스트, columns 2개는 리스트
```
