```python
import pandas as pd
df = pd.read_excel("file/score.xlsx")
df
```

```python
df.replace({'구로고':'신림고'},inplace=True)
```

```python
df.replace({'구로고':'신림고','디지털고':'포켓몬고'},inplace=True)
```

```python
# 컬럼 추가 : 없는 컬럼이름과 값을 입력하면, 컬럼이 추가됨


df['결과'] = 0
df
```

```python
# 컬럼 값의 수정 : replace로 변경해도 되고, 있는 컬럼에 값을 입력하면 수정됨


df['결과'] = "불합격"
df
```

```python
# 합계를 추가하려면..

df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['과학'] + df['사회']
df
```

```python
# 조건으로 컬럼 수정 - 조건에 맞는 컬럼의 값을 변경할 수 있음
# 컬럼에서 전체 수정은 replace, 일부만 수정은 loc
# 합계가 400이상인 경우
```

```python
# [ 1.글자수정 - replace() ]
# columns 수정 - replace()를 이용하여 컬럼의 값을 변경할 수 있음.
# df.replace({'구로고': '신림고'}, inplace=True)  # 학교이름변경
# df['학교'] = df['학교'].replace({'구로고': '신림고','디지털고':'포켓몬고'} )  # 학교이름변경
# df.replace({'구로고': '신림고','디지털고':'포켓몬고'},inplace=True )  # 학교이름변경
#--------------------------
# [ 2.컬럼전체수정 - 컬럼에 값을 입력 ]
# 존재하는 column에 값을 입력하면 수정이 됨.
#  예) df['합계']가 존재할때, 다시 값을 입력하면 수정이 됨. df['합계'] = 1
#--------------------------
# [ 3.부분수정 - loc ]
# 조건으로 컬럼 수정 - 조건에 맞는 컬럼의 값을 변경할 수 있음.
# df.loc[df['합계']>400, '결과'] = '합격'
#--------------------------
# columns 추가 - 없는 컬럼을 입력하면 추가가 됨.
# df['결과'] = 0
# df['결과'] = "불합격"
# df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['과학'] + df['사회']
```
