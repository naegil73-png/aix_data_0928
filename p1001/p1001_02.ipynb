```python
import pandas as pd
df = pd.read_excel("file/students.xlsx",index_col="번호")
df
```

```python
# 합계, 평균, 결과 컬럼을 추가
# 결과는 평균 점수가 70점 이상이면 합격, 미만은 불합격으로 출력, 합격과 불합격 빈도도 출력
```

```python
df['합계'] = df['국어'] + df['영어'] + df['수학']
df
```

```python
df['평균'] = (df['국어'] + df['영어'] + df['수학'])/3
df
```

```python
df['결과'] = "불합격"
df
```

```python
df.loc[df['평균']>=70,'결과'] = "합격"
df
```
