```python
def cal(x):
    return x+2

# (lambda x:x+2) 이렇게 표현 가능(1줄 함수)
```

```python
# 1. 함수사용 - 천단위 표시
def cal_change(x):
    return  f"{x:,}"
    # return  '{:,d}'.format() # 숫자를 문자변형

df['관객 수'] = df['관객 수'].apply(cal_change)

#  2. lambda함수적용
df['관객 수'] = df['관객 수'].apply(lambda x: f"{x:,}")

#  3. map사용
# df['관객 수'] = df['관객 수'].map('{:,}'.format)
```
