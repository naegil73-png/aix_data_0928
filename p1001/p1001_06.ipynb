```python
import pandas as pd
df = pd.read_excel('file/score.xlsx',index_col="지원번호")
df
```

```python
# 부류별 그룹핑 groupby : size, mean, sum, max, min
df.groupby('학교').get_group('디지털고') # 학교를 기준으로 그룹핑해서 디지털고에 대당하는 것을 출력하라
df.groupby('학교').size() # 학교를 기준으로 그룹핑해서 각각의 크기를 알려줌
df.groupby('학교').size()['디지털고'] # 학교를 기준으로 그룹핑해서 디지털고 크기를 알려줌
# df['학교'].unique()
df.groupby('학교')['키'].mean() # 학교기준으로 학교별 키 평균
# df.groupby('학교').mean() # 학교 전체는 안나옴
df.groupby('학교')['국어'].mean()
df.groupby('학교')[['국어','영어','수학']].mean()
df.groupby('학교')[['국어','영어','수학']].sum()
df.groupby('학교')[['국어','영어','수학']].max()
```

```python
df.groupby('학교').size() # 위, 아래는 완전히 같은 명령문임
df['학교'].value_counts 
```

```python
df['학년'] = [3,3,2,1,1,3,2,2]
df
```

```python
# df.groupby('학교')['학년'].value_counts().loc['구로고']
df.groupby('학교')['학년'].value_counts().loc['디지털고']
df.groupby('학교')['학년'].value_counts()
```

```python
df.groupby('학교')['학년'].value_counts(normalize=True).loc['디지털고'] # normalize는 백분율화함
```

```python
df.groupby('학년')['키'].mean()
df.groupby('학년')['키'].mean().sort_values()
```

```python
df.groupby('학년')[['국어','영어']].mean()
df.groupby('학년')[['국어','영어']].sum().sort_values('국어') # 두개의 요소가 있으면, 어떤 기준으로 정렬할 것인지 표시해야 함
df.groupby('학교')[['국어','영어']].count().sort_values('영어')
```
