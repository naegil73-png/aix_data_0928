```python
# 1. item_name 당 주문 개수를 출력하시오. 10개만 출력
# 2. item_name 당 주문 총량(quantity)을 출력하시오.
# 3. 가장 많이 주문한 item_name  10개를 출력하시오.
# 4. 주문당 평균 금액을 출력하시오.
# 5. 한 주문에 10달러 이상 사용한 주문 번호(id) 출력하시오.
```

```python
import pandas as pd
df = pd.read_csv("file/chipotle.tsv",sep="\t")
df
```

```python
# 먼적, info, describe
df.info()
df.describe()
```

```python
# 1) 
df['order_id'].unique()
item = df.groupby('item_name')
item.astype(int)
top_item = item.sort_values(ascending=False)
top_item[:5]
```

```python

# 3)
df.groupby('item_name')['order_id'].count().sort_values(ascending=False)[:10]
# df['item_name].value_counts().head(10) 해도 됨
```

```python
# 2)
df['quantity'] = df['quantity'].astype(int)
item = df.groupby('item_name').size() # item = df.groupby('item_name')['quantity]].sum().sort_values(ascending=False)
item.sort_values(ascending=False)
```

```python
# 2)
df['quantity'].astype(int)
df['item_price'] = df['item_price'].str.replace("$",'')
df['item_price'] = df['item_price'].str.replace(",",'').astype(float)
df['item_price']
df['item_price'].sum()
```

```python
df['total_cost'] = df['item_price']*df['quantity']
df.groupby('item_name')
```

```python
# 5)
df[df['total_cost']>=50.00]['order_id']
```

```python
# 4)
df.groupby('order_id')['item_price'].sum().mean()
```

```python
# Chicken Bowl을 2개 이상 구매한 주문서 출력
filt = (df['item_name']=="Chicken Bowl") & (df['quantity']>=2)
df[filt]
```
