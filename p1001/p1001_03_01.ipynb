```python
import pandas as pd
df = pd.read_excel("file/score.xlsx",index_col="지원번호")
df
```

```python
# 컬럼 전체 이름변경
# df['학교'] = df['학교']+'등학교'
df.replace({'고':'등학교'},inplace=True)
df['학교'] = df['학교'].str.replace({'고':'고등학교'})
df['학교']
```

```python
# 키 + cm추가
# df['키'] = df['키'] + 'cm' # 오류 남 -> 키의 수는 숫자형, cm는 문자형 -> 다른 형태로 오류남
df['키'] = df['키'] + 100 # 이건 가능(숫자형 + 숫자형)
df
```

```python
# 키 + cm추가 하려면, 키를 문자형으로 변환해야 함
df['키'] = df['키'].astype(str)
df.info
```

```python
# 함수 적용 : apply() - 컬럼 전체에 적용
def add_cm(h):
    return str(h)+'센티미터'

df['키'] = df['키'].apply(add_cm)
df
```

```python
# 함수적용

def capChange(lang):
    pass


# df['SW특기'].str.lower()

def capChange(lang):
    if pd.notnull(lang):
        return lang.lower()
    else:
        return "특기없음"

df['SW특기'] = df['SW특기'].apply(capChange)
df['SW특기']
```

```python
# 문제 : 
# 1.합계 컬럼(과목 총점 계산해서)을 추가하시오.
# 2.평균 컬럼을 추가하시오.
# 3.등급 컬럼을 추가해서 90점 이상 A, 80점 이상 B, 70점 이상 C, 60점 이상 D, 그 외는 F로 입력하시오.
# 3번은 함수를 사용해서 처리
```

```python
df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['과학'] + df['사회']
df['평균'] = df['합계'] / 5
df['등급'] = "F"
df
```

```python

df['평균'] = df['평균'].astype(int)
def grade(평균):
    if 평균 >=90: return "A"
    elif 80<= 평균 < 90: return "B"
    elif 70<= 평균 < 80: return "C"
    elif 60<= 평균 < 70: return "D"
    else: return "F"

df['등급'] = df['평균'].apply(grade)
df
```

```python
# rank 변수
df['합계'].rank() # 합계를 기준으로 등수를 매겨라 -> 낮은 점수가 낮은 숫자로 나타남 -> 높을 수록 높은 등수가 나오려면, ascending = False넣어야 함
```

```python
df['등수'] = df['합계'].rank(ascending=False).astype(int)
df
```
